```python
def roi(revenue, cost):
  assert cost != 0, "Invalid cost"
  assert isinstance(revenue, (int, float))
  assert isinstance(cost, (int, float))
  return (revenue - cost) / cost
```

```python
def loan_payment(principal, annual_rate, months):
  assert isinstance(principal, (int, float))
  assert isinstance(annual_rate, (int, float))
  assert isinstance(months, int)

  if principal <= 0 or annual_rate < 0 or months <= 0:
    raise ValueError("Invalid input values")

  try:
    r = annual_rate / 12
    if r == 0:
      pmt = principal / months
    else:
      pmt = r * principal / (1 - (1 + r) ** (-months))
  except Exception as e:
    raise Exception(f"Error: {e}")

  return pmt
```

```python
class Greeting:
  def __init__(self, first, last):
    self.first = first
    self.last = last

  def display(self, formal=False):
    if formal:
      print(f"{self.last}, {self.first}")
    else:
      print(f"{self.first}")

  def __str__(self):
    return "Class Greeting\nReturns formal or informal greetings"

s = Greeting("John", "Doe")
print(s)
s.display()
s.display(formal=True)
```

```python
class Student:
  def __init__(self, name, number):
    self.__name = name
    self.number = number
    self.__courses = []

  def get_courses(self):
    return self.__courses

  def get_name(self):
    return self.__name

  def set_name(self, newname):
    self.__name = newname

  def __len__(self):
    return len(self.__courses)

  def enroll(self, course):
    if course in self.__courses:
      print(f"Student {self.__name} is already enrolled in {course}.")
    else:
      self.__courses.append(course)
```

```python
s = Student("Jane Doe", 12345)
print("Student name:", s.get_name())
s.set_name("Jane B. Doe")
print("Student name:", s.get_name())
print("Courses enrolled in:", s.get_courses())
s.enroll('BTE320')
s.enroll('BTE320')
print(f"Student {s.get_name()} has enrolled in {len(s)} course(s): {s.get_courses()}")
```

**In Class Assignment #15 (4-13-2026)**

```python
class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def distance(self, other):
    dx = (self.x - other.x) ** 2
    dy = (self.y - other.y) ** 2
    dist = (dx + dy) ** 0.5
    return dist

class Rectangle:
  def __init__(self, x1, y1, x2, y2, x3, y3, x4, y4):
    self.p1 = Point(x1, y1) #upper left
    self.p2 = Point(x2, y2) #upper right
    self.p3 = Point(x3, y3) #lower left
    self.p4 = Point(x4, y4) #lower right
    self.width = self.p1.distance(self.p2)
    self.height = self.p1.distance(self.p3)

  def area(self):
    return self.width * self.height
```

```python
square = Rectangle(0,3,3,3,0,0,3,0)
square.area()
```

**In Class Assignment #16 (4-20-2026)**

```python
class Employee:
  def __init__(self, name, wage, hours_worked):
    self.name = name
    self.__wage = wage
    self.hours_worked = hours_worked

  def calculate_pay(self):
    return self.__wage * self.hours_worked

  def display_info(self):
    print(f"Name: {self.name}")
    print(f"Hours Worked: {self.hours_worked}")
    print(f"Weekly Pay: {self.calculate_pay()}")

  def get_wage(self):
    return self.__wage

  def set_wage(self, new_wage):
    if new_wage >= 0:
      self.__wage = new_wage
    else:
      print("Invalid wage. Wage cannot be negative.")
```

```python
class Manager(Employee):
  def __init__(self, name, wage, hours_worked, bonus):
    super().__init__(name, wage, hours_worked)
    self.bonus = bonus

  def calculate_pay(self):
    return super().calculate_pay()+self.bonus

emp1 = Employee("Alice", 20, 35)
mgr1 = Manager("Bob", 30, 40, 200)
```

```python
print("EMPLOYEE")
emp1.display_info()
print("\nMANAGER")
mgr1.display_info()
print(f"Bonus: {mgr1.bonus}")
print(f"Total Pay: {mgr1.calculate_pay()}")
```
