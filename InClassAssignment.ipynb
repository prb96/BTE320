```python
sales = float(input("Enter sales ($): "))
tax_rate = float(input("Enter tax rate (%): "))
print("Sales:", sales)
cogs= 0.4 * sales
print("Cost of Goods Sold:", cogs)
gp = sales - cogs
print ("Gross Profit:", gp)
taxes = (tax_rate/100)*gp
print("Taxes:",taxes)
Net_income = gp - taxes
print("Net Income:", Net_income)
```
