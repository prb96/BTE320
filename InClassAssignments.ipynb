**In Class Assignment #1 (1-26-2026)**

```python
sales = float(input("Enter sales ($): "))
tax_rate = float(input("Enter tax rate (%): "))
print("Sales:", sales)
cogs= 0.4 * sales
print("Cost of Goods Sold:", cogs)
gp = sales - cogs
print ("Gross Profit:", gp)
taxes = (tax_rate/100)*gp
print("Taxes:",taxes)
Net_income = gp - taxes
print("Net Income:", Net_income)
```

**In Class Assignment #2 (2-2-2026)**

```python
number1 = float(input("Enter first number: "))
operator = input("Enter operator: ")
number2 = float(input("Enter second number: "))

if operator == "+":
  res = number1 + number2
elif operator == "-":
  res = number1 - number2
elif operator == "*":
  res = number1 * number2
elif operator == "**":
  res = number1 ** number2
elif operator == "/":
  res = number1 / number2
elif operator == "//":
  res = number1 // number2
elif operator == "%":
  res = number1 % number2
else:
  print("Invalid operator")
  res = None

print(f"Result: {res}")
```
